### GroupBy (Aggregations)

Examples:

- Total Amount by region  
  `df.groupby("region")["Amount"].sum()`

- Count orders by product  
  `df.groupby("product")["order_id"].count()`

- Avg quantity by region  
  `df.groupby("region")["quantity"].mean()`

- Max amount by product  
  `df.groupby("product")["Amount"].max()`


```python
df.groupby("region")["Amount"].sum()
```

```python
df.groupby("product")["order_id"].count()
```

```python
df.groupby("region")["quantity"].mean()
```

```python
df.groupby("product")["Amount"].max()
```

### Joining Two CSV Files

Steps:
1. Load sales.csv  
2. Load customers.csv  
3. Merge on `customer_id` using LEFT JOIN


```python
import pandas as pd
df_sales=pd.read_csv("../../datasets/sales.csv")
df_customers=pd.read_csv("../../datasets/customers.csv")
df_sales["Amount"]=df_sales["price"] * df_sales["quantity"]                         
df_merged=df_sales.merge(df_customers, on="customer_id",how="left") 
df_merged.head()
```

### Highest Value Order (Top 1)



```python
df_merged.sort_values("Amount", ascending=False).head(1)
```
